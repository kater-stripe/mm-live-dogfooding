```python
import requests
import json
from datetime import datetime
from typing import Dict, Any

# ============================================================================
# CONFIGURATION - UPDATE THESE VALUES
# ============================================================================

# Your LIVE secret key
SECRET_KEY = ""  #kater key REPLACE THIS

# Platform account ID (will be populated after first API call if needed)
PLATFORM_ACCOUNT_ID = "acct_1JmyhCGPC58OcFYO"  # Test Connect vg@stripe.com

# API version for v2 endpoints
API_VERSION = "2025-09-30.preview"

# Test data storage
test_data = {
    "connected_account_id": "acct_1SzGCz8rmri64Y6r",
    "financial_account_id": "fa_65U8eHo1DO4IHeUArqA16U8eAL8GE9WV3GpLWntHBDE6Pw",
    "financial_address_id": "finaddr_61U8eke1cqc1T1xNf16U8eAL8GE9WV3GpLWntHBDEQqm",
    "financial_account_id_2": "fa_65U8hVIAeDKurgIwUp416U8eAL8GE9WV3GpLWntHBDESMK",
     # Kate Roumruk cardholder
    "cardholder_id": "ich_1SzIzV8rmri64Y6rnksaKQ2a",
    "card_id": "ic_1SzJ0W8rmri64Y6r6W9A84c6",
    # Kate's Monzo bank attached to Recipient
    "recipient_account_id": "acct_1SzJ9K8rmr2k2zYk", # A recipient on acct_1SzGCz8rmri64Y6r
    "payout_method_id": "gbba_61U8h8nyuiGS94AVH16U8h6g9NE9qWlDDYP9bb5WKF04",
    "connected_account_id_2": None,  # For testing CA-to-CA transfers
    "card_program_id": "iprg_1SEDyVGPC58OcFYO2iM8vkUH",
    "recipient_account_id_us": None,  # For cross-border testing
    "payout_method_id_us": None,
    "bank_account_id": "gbba_61U8hXhIWVHgAcxnL16U8eAL8GE9WV3GpLWntHBDEJlY",  # Kate's Monzo Bank account
}

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def print_section(title: str):
    """Print a formatted section header"""
    print("\n" + "="*80)
    print(f"  {title}")
    print("="*80 + "\n")

def print_result(response: requests.Response, show_full: bool = False):
    """Print API response in a readable format"""
    print(f"Status: {response.status_code}")
    
    if response.status_code >= 400:
        print("❌ ERROR")
        print(json.dumps(response.json(), indent=2))
        return None
    
    print("✅ SUCCESS")
    data = response.json()
    
    if show_full:
        print(json.dumps(data, indent=2))
    else:
        # Print key fields only
        if 'id' in data:
            print(f"ID: {data['id']}")
        if 'object' in data:
            print(f"Object: {data['object']}")
        if 'status' in data:
            print(f"Status: {data['status']}")
        if 'balance' in data:
            print(f"Balance: {data['balance']}")
    
    return data

def v2_headers(stripe_account: str = None, stripe_context: str = None) -> Dict[str, str]:
    """Generate headers for v2 API calls"""
    headers = {
        "Authorization": f"Bearer {SECRET_KEY}",
        "Stripe-Version": API_VERSION,
        "Content-Type": "application/json"
    }
    if stripe_account:
        headers["Stripe-Account"] = stripe_account
    if stripe_context:
        headers["Stripe-Context"] = stripe_context
    return headers

def v1_auth() -> tuple:
    """Generate auth tuple for v1 API calls"""
    return (SECRET_KEY, "")

print("✅ Setup complete!")
print(f"API Version: {API_VERSION}")
print(f"Secret Key: {SECRET_KEY[:15]}...")
```

# Step 2A: Create v2 Connected Account with Storer Configuration

## 🎯 Objective
Create a new Connected Account (your customer/merchant) with all necessary configurations for v2 Financial Accounts and Issuing.

## 🔍 What This Step Does
Creates a v2 Connected Account with four key configurations:
1. **Storer**: Enables v2 Financial Accounts (money storage, transfers, payments)
2. **Merchant**: Enables payment acceptance (cards, bank transfers, etc.)
3. **Card Creator**: Enables Issuing cards (commercial prepaid cards)
4. **Recipient**: Enables receiving payouts from platform

## 📋 Key Configuration Details

**Storer Capabilities:**
- `holds_currencies`: GBP storage
- `financial_addresses`: Bank account details for receiving transfers
- `inbound_transfers`: Receive money from external banks
- `outbound_transfers`: Send money to own FAs or external banks
- `outbound_payments`: Send money to recipients or other CAs

**Card Creator Capabilities:**
- `commercial.stripe.prepaid_card`: Issue prepaid cards

**Merchant Capabilities:**
- Card payments, BACS, GB bank transfers, etc.

## ✅ Expected Outcome
- New Connected Account created with ID starting with `acct_`
- Account status will likely be `pending` or `requires_onboarding`
- All four configurations should be present
- Account ID will be stored in `test_data['connected_account_id']`

## 📝 What to Document
- Account creation time (should be < 3 seconds)
- Account ID generated
- Initial account status
- Any errors or warnings in the response

# Step 5: Money Movement - Outbound Payments to Recipient

```python
print_section("Step 5A: Create Recipient Account (UK)")

if not test_data['connected_account_id']:
    print("⚠️ No connected account ID.")
else:
    payload = {
        "contact_email": "recipient-uk@test.com",
        "display_name": "UK Recipient",
        "identity": {
            "country": "GB",
            "entity_type": "individual",
            # "individual": {
            #     "given_name": "Jenny",
            #     "surname": "Recipient"
            # }
        },
        "include": [
            "requirements",
            "configuration.recipient",
            "identity"
        ],
        "configuration": {
            "recipient": {
                "capabilities": {
                    "bank_accounts": {
                        "local": {"requested": True}
                    }
                }
            }
        }
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/core/accounts",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    recipient = print_result(response, show_full=False)
    
    if recipient:
        test_data['recipient_account_id'] = recipient['id']
        print(f"\n👤 Recipient Account ID: {test_data['recipient_account_id']}")
```

```python
print_section("Step 5B: Create Account Link for Recipient Onboarding")

if not test_data['recipient_account_id']:
    print("⚠️ No recipient account ID. Run Step 5A first.")
else:
    payload = {
        "account": test_data['recipient_account_id'],
        "use_case": {
            "type": "account_onboarding",
            "account_onboarding": {
                "collection_options": {
                    "fields": "currently_due",
                    "future_requirements": "include"
                },
                "configurations": ["recipient"],
                "refresh_url": "https://example.com/refresh",
                "return_url": "https://example.com/return"
            }
        }
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/core/account_links",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    link = print_result(response, show_full=False)
    
    if link:
        print(f"\n🔗 Recipient Onboarding URL: {link['url']}")
        print("\n⚠️ Complete the onboarding at the URL above before proceeding.")
```

```python
print_section("Step 5C: Get Recipient Payout Method")

if not test_data['recipient_account_id']:
    print("⚠️ No recipient account ID.")
elif not test_data['connected_account_id']:
    print("⚠️ No connected account ID.")
else:
    # Note: Stripe-Context header uses path format: platform_account/recipient_account
    stripe_context = f"{test_data['connected_account_id']}/{test_data['recipient_account_id']}"
    
    response = requests.get(
        "https://api.stripe.com/v2/money_management/payout_methods",
        headers=v2_headers(stripe_context=stripe_context)
    )
    
    methods = print_result(response, show_full=False)
    
    if methods and methods.get('data'):
        test_data['payout_method_id'] = methods['data'][0]['id']
        print(f"\n💳 Payout Method ID: {test_data['payout_method_id']}")
        print(f"Type: {methods['data'][0].get('type', 'N/A')}")
    else:
        print("⚠️ No payout methods found. Complete recipient onboarding first.")
```

```python
print_section("Step 5D: Make Outbound Payment to Recipient")

if not test_data['financial_account_id']:
    print("⚠️ No financial account ID.")
elif not test_data['payout_method_id']:
    print("⚠️ No payout method ID. Complete Steps 5A-5C first.")
else:
    payload = {
        "amount": {
            "value": 100,  # £100.00
            "currency": "gbp"
        },
        "from": {
            "currency": "gbp",
            "financial_account": test_data['financial_account_id']
        },
        "to": {
            "currency": "gbp",
            "payout_method": test_data['payout_method_id'],
            "recipient": test_data['recipient_account_id']
        },
        "description": "Live test payment to recipient"
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/money_management/outbound_payments",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    payment = print_result(response, show_full=False)
    
    if payment:
        print(f"\n✅ Outbound Payment Created!")
        print(f"Payment ID: {payment['id']}")
        print(f"Status: {payment.get('status', 'N/A')}")
```

# Step 6: Outbound Transfers (Me-to-Me)

```python
print_section("Step 6A: Create Second Financial Account")

if not test_data['connected_account_id']:
    print("⚠️ No connected account ID.")
else:
    payload = {
        "type": "storage",
        "storage": {
            "holds_currencies": ["gbp"]
        },
        "display_name": "Secondary Account"
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/money_management/financial_accounts",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    fa2 = print_result(response, show_full=False)
    
    if fa2:
        test_data['financial_account_id_2'] = fa2['id']
        print(f"\n💰 Second FA ID: {test_data['financial_account_id_2']}")
```

```python
print_section("Step 6B: Transfer Between Own FAs (Instant)")

if not test_data['financial_account_id']:
    print("⚠️ No source financial account ID.")
elif not test_data['financial_account_id_2']:
    print("⚠️ No destination financial account ID. Run Step 6A first.")
else:
    payload = {
        "amount": {
            "value": 100,  # £50.00
            "currency": "gbp"
        },
        "from": {
            "currency": "gbp",
            "financial_account": test_data['financial_account_id']
        },
        "to": {
            "currency": "gbp",
            "payout_method": test_data['financial_account_id_2']
        },
        "description": "Transfer to secondary FA"
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/money_management/outbound_transfers",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    transfer = print_result(response, show_full=False)
    
    if transfer:
        print(f"\n✅ Transfer Created!")
        print(f"Transfer ID: {transfer['id']}")
        print(f"Status: {transfer.get('status', 'N/A')}")
```

```python
print_section("Step 6C: Create External Bank Account (UK)")

if not test_data['connected_account_id']:
    print("⚠️ No connected account ID.")
else:
    payload = {
        "account_number": "57229587",  #Kate's Monzo Bank Account
        "sort_code": "040004",
        "confirmation_of_payee": {
            "initiate": True
        }
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/core/vault/gb_bank_accounts",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    bank_account = print_result(response, show_full=False)
    
    if bank_account:
        test_data['bank_account_id'] = bank_account['id']
        print(f"\n🏦 Bank Account ID: {test_data['bank_account_id']}")
```

```python
print_section("Step 6D: Acknowledge Confirmation of Payee (CoP)")

if not test_data['bank_account_id']:
    print("⚠️ No bank account ID. Run Step 6C first.")
else:
    response = requests.post(
        f"https://api.stripe.com/v2/core/vault/gb_bank_accounts/{test_data['bank_account_id']}/acknowledge_confirmation_of_payee",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json={}
    )
    
    result = print_result(response, show_full=False)
    
    if result:
        print("\n✅ CoP Acknowledged!")
```

```python
print_section("Step 6E: Transfer to External Bank Account")

if not test_data['financial_account_id']:
    print("⚠️ No financial account ID.")
elif not test_data['bank_account_id']:
    print("⚠️ No bank account ID. Complete Steps 6C-6D first.")
else:
    payload = {
        "amount": {
            "value": 2000,  # £20.00
            "currency": "gbp"
        },
        "from": {
            "currency": "gbp",
            "financial_account": test_data['financial_account_id']
        },
        "to": {
            "currency": "gbp",
            "payout_method": test_data['bank_account_id']
        },
        "description": "Transfer to external bank"
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/money_management/outbound_transfers",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    transfer = print_result(response, show_full=False)
    
    if transfer:
        print(f"\n✅ Transfer to External Bank Created!")
        print(f"Transfer ID: {transfer['id']}")
        print(f"Status: {transfer.get('status', 'N/A')}")
```

# Step 7: Cross-Border Outbound Payment (GBP → USD)

```python
print_section("Step 7A: Create US Recipient Account")

if not test_data['connected_account_id']:
    print("⚠️ No connected account ID.")
else:
    payload = {
        "contact_email": "recipient-us@test.com",
        "display_name": "US Recipient",
        "identity": {
            "country": "US",
            "entity_type": "individual",
            "individual": {
                "given_name": "John",
                "surname": "Recipient"
            }
        },
        "include": [
            "requirements",
            "configuration.recipient",
            "identity"
        ],
        "configuration": {
            "recipient": {
                "capabilities": {
                    "bank_accounts": {
                        "local": {"requested": True}
                    }
                }
            }
        }
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/core/accounts",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    recipient_us = print_result(response, show_full=False)
    
    if recipient_us:
        test_data['recipient_account_id_us'] = recipient_us['id']
        print(f"\n👤 US Recipient Account ID: {test_data['recipient_account_id_us']}")
```

```python
print_section("Step 7B: Onboard US Recipient (Get Bank Details)")

if not test_data['recipient_account_id_us']:
    print("⚠️ No US recipient account ID. Run Step 7A first.")
else:
    payload = {
        "account": test_data['recipient_account_id_us'],
        "use_case": {
            "type": "account_onboarding",
            "account_onboarding": {
                "collection_options": {
                    "fields": "currently_due",
                    "future_requirements": "include"
                },
                "configurations": ["recipient"],
                "refresh_url": "https://example.com/refresh",
                "return_url": "https://example.com/return"
            }
        }
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/core/account_links",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    link = print_result(response, show_full=False)
    
    if link:
        print(f"\n🔗 US Recipient Onboarding URL: {link['url']}")
        print("\n⚠️ Complete the onboarding before proceeding.")
```

```python
print_section("Step 7C: Get US Recipient Payout Method")

if not test_data['recipient_account_id_us']:
    print("⚠️ No US recipient account ID.")
else:
    stripe_context = f"{test_data['connected_account_id']}/{test_data['recipient_account_id_us']}"
    
    response = requests.get(
        "https://api.stripe.com/v2/money_management/payout_methods",
        headers=v2_headers(stripe_context=stripe_context)
    )
    
    methods = print_result(response, show_full=False)
    
    if methods and methods.get('data'):
        test_data['payout_method_id_us'] = methods['data'][0]['id']
        print(f"\n💳 US Payout Method ID: {test_data['payout_method_id_us']}")
```

```python
print_section("Step 7D: Setup Outbound Payment Intent")

if not test_data['payout_method_id_us']:
    print("⚠️ No US payout method. Complete Steps 7A-7C first.")
else:
    stripe_context = f"{test_data['connected_account_id']}/{test_data['recipient_account_id_us']}"
    
    payload = {
        "payout_method": test_data['payout_method_id_us'],
        "usage_intent": "payment"
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/money_management/outbound_setup_intents",
        headers=v2_headers(stripe_context=stripe_context),
        json=payload
    )
    
    setup = print_result(response, show_full=False)
    
    if setup:
        print("\n✅ Outbound Setup Intent Created!")
```

```python
print_section("Step 7E: Get FX Quote")

if not test_data['financial_account_id']:
    print("⚠️ No financial account ID.")
elif not test_data['payout_method_id_us']:
    print("⚠️ No US payout method. Complete Steps 7A-7D first.")
else:
    payload = {
        "amount": {
            "value": 10000,  # £100.00
            "currency": "gbp"
        },
        "from": {
            "currency": "gbp",
            "financial_account": test_data['financial_account_id']
        },
        "to": {
            "currency": "usd",
            "payout_method": test_data['payout_method_id_us'],
            "recipient": test_data['recipient_account_id_us']
        }
    }
    
    response = requests.post(
        "https://api.stripe.com/v2/money_management/outbound_payment_quotes",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        json=payload
    )
    
    quote = print_result(response, show_full=True)
    
    if quote:
        print("\n💱 FX Quote Obtained!")
        print(f"Quote ID: {quote['id']}")
        if 'exchange_rate' in quote:
            print(f"Exchange Rate: {quote['exchange_rate']}")
```

# Step 8: Review & Monitor

```python
print_section("Final Balance Check")

if test_data['financial_account_id']:
    response = requests.get(
        f"https://api.stripe.com/v2/money_management/financial_accounts/{test_data['financial_account_id']}",
        headers=v2_headers(stripe_account=test_data['connected_account_id'])
    )
    
    fa = print_result(response, show_full=False)
    
    if fa and 'balance' in fa:
        print("\n💰 Final Balance Summary:")
        balance = fa['balance']
        for balance_type, currency_amounts in balance.items():
            if isinstance(currency_amounts, dict):
                for currency, amount_data in currency_amounts.items():
                    if isinstance(amount_data, dict) and 'value' in amount_data:
                        value = amount_data['value']
                        symbol = '£' if currency == 'gbp' else '$' if currency == 'usd' else currency.upper()
                        print(f"  {currency.upper()} {balance_type}: {symbol}{value/100:.2f}")
```

```python
print_section("Transaction History")

if test_data['financial_account_id']:
    response = requests.get(
        "https://api.stripe.com/v2/money_management/transactions",
        headers=v2_headers(stripe_account=test_data['connected_account_id']),
        params={"financial_account": test_data['financial_account_id']}
    )
    
    txns = print_result(response, show_full=False)
    
    if txns and txns.get('data'):
        print(f"\n📋 Recent Transactions ({len(txns['data'])}):\n")
        for txn in txns['data'][:10]:  # Show first 10
            print(f"  {txn['id']}")
            if 'amount' in txn:
                amt = txn['amount']
                symbol = "£" if amt.get('currency') == 'gbp' else "$"
                print(f"    Amount: {symbol}{amt.get('value', 0)/100:.2f}")
            print(f"    Type: {txn.get('type', 'N/A')}")
            print()
```
